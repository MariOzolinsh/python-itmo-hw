# Домашнее задание: класс `Chocolate`

Реализация класса `Chocolate` с проверкой корректности цены, методом `show_my_chocolate`, переопределённым методом `__str__` и демонстрацией работы.

## 1. Реализация класса `Chocolate`

```python
from typing import Optional


class Chocolate:
    """Класс для представления шоколадного батончика."""

    def __init__(self, price: float, additive: Optional[str] = None):
        if price <= 0:
            raise ValueError("Цена батончика должна быть больше нуля")
        self.price = float(price)
        self.additive = additive

    def show_my_chocolate(self):
        if self.additive:
            print(f"Шоколадный батончик, добавка: {self.additive}")
        else:
            print("Обычный шоколадный батончик")

    def __str__(self):
        return f"Шоколадный батончик, {self.price} рублей"
```

## 2. Демонстрация работы с объектами класса

```python
# 1. Батончик с добавкой
choco_nuts = Chocolate(125.5, "лесной орех")
print(choco_nuts)
choco_nuts.show_my_chocolate()

print()

# 2. Обычный батончик (без указания добавки)
choco_classic = Chocolate(89.9)
print(choco_classic)
choco_classic.show_my_chocolate()

print()

# 3. Батончик с передачей именованных аргументов
choco_caramel = Chocolate(price=150.0, additive="соленая карамель")
print(choco_caramel)
choco_caramel.show_my_chocolate()
```

## 3. Проверка обработки некорректной цены (`ValueError`)

```python
# 4. Проверка исключения ValueError при цене <= 0
for invalid_price in (0.0, -45.0):
    try:
        bad_choco = Chocolate(invalid_price, "изюм")
    except ValueError as e:
        print(f"Цена {invalid_price} отклонена — ValueError: {e}")
```
